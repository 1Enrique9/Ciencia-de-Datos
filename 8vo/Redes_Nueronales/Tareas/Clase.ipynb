```python
import numpy as np

def f(x):
    return np.sin(x) * np.exp(-x**2)

# Derivada exacta en x=1: cos(1)*exp(-1) + sin(1)*(-2)*exp(-1)
x = 1.0
exact = np.cos(x) * np.exp(-x**2) + np.sin(x) * (-2*x) * np.exp(-x**2)

print(f"{'h':>12} | {'Approx':>18} | {'Error':>15}")
print("-" * 52)

for h in [1e-1, 1e-4, 1e-7, 1e-10, 1e-14]:
    approx = (f(x + h) - f(x - h)) / (2 * h)
    error = abs(approx - exact)
    print(f"{h:>12.0e} | {approx:>18.10f} | {error:>15.2e}")
```
