```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Cargar datos
data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)

# Centrar datos
X_centered = X - X.mean()

# Matriz de covarianza 
C = np.cov(X_centered, rowvar=False)

C
```

# Método 1: autovalores/autovectores

```python
eigvals, eigvecs = np.linalg.eig(C)

# Ordenar de mayor a menor
idx = np.argsort(eigvals)[::-1]
eigvals_sorted = eigvals[idx]
eigvecs_sorted = eigvecs[:, idx]

eigvals_sorted, eigvecs_sorted
```

# Método 2: PCA mediante SVD

```python
U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)

# Varianzas por componente
S2 = (S**2) / (X.shape[0] - 1)

# Los vectores principales están en Vt
V_svd = Vt.T

S2, V_svd
```

# Comparacion 

```python
np.allclose(eigvals_sorted, S2)
```

```python
np.allclose(np.abs(eigvecs_sorted), np.abs(V_svd))
```

Podemos que Ambos métodos producen la misma matriz de componentes principales

SVD y autovalores/autovectores son equivalentes para PCA.

Los autovalores coinciden:  4.228,0.242,0.078,0.023

Las matrices de autovectores son idénticas salvo por signo

Lo cual es normal.
